# StockGuard EDA
수집된 유튜브 데이터 탐색적 분석

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
from pathlib import Path

DB_PATH = Path('../data/youtube_channels.db')
conn = sqlite3.connect(DB_PATH)
conn.row_factory = sqlite3.Row
```

## 1. 데이터 로드

```python
df_channels = pd.read_sql('SELECT * FROM channels', conn)
df_videos   = pd.read_sql('SELECT * FROM videos', conn)
df_comments = pd.read_sql('SELECT * FROM comments', conn)

print(f'채널  : {len(df_channels):,}개')
print(f'영상  : {len(df_videos):,}개')
print(f'댓글  : {len(df_comments):,}개')
```

## 2. 기본 통계

```python
df_channels.describe()
```

```python
df_videos.describe()
```

## 3. 결측치 확인

```python
print('=== channels ===')
print(df_channels.isnull().sum())
print('\n=== videos ===')
print(df_videos.isnull().sum())
print('\n=== comments ===')
print(df_comments.isnull().sum())
```

## 4. 분포 시각화

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

df_channels['subscriber_count'].clip(upper=df_channels['subscriber_count'].quantile(0.99)).hist(ax=axes[0], bins=50)
axes[0].set_title('구독자 수 분포')

df_videos['view_count'].clip(upper=df_videos['view_count'].quantile(0.99)).hist(ax=axes[1], bins=50)
axes[1].set_title('영상 조회수 분포')

df_videos['duration_seconds'].clip(upper=df_videos['duration_seconds'].quantile(0.99)).hist(ax=axes[2], bins=50)
axes[2].set_title('영상 길이 분포 (초)')

plt.tight_layout()
plt.show()
```

## 5. 검색어별 채널 수

```python
df_channels['search_query'].value_counts().plot(kind='barh', figsize=(8, 10))
plt.title('검색어별 수집 채널 수')
plt.tight_layout()
plt.show()
```
